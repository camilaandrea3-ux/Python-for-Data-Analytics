```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
noise_df = df[df['Complaint Type'].str.contains('Noise', case=False, na=False)]
```

```python
borough_counts = noise_df['Borough'].value_counts()
```

```python
plt.figure(figsize=(10, 6))
borough_counts.sort_values().plot(kind='barh', color='steelblue')

plt.title('Number of Noise Complaints by Borough', fontsize=14)
plt.xlabel('Count')
plt.ylabel('Borough')
plt.grid(axis='x', alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
df['Complaint Type'].value_counts().plot(
    kind='bar',
    figsize=(10,5),
    title='Visualization 2: Complaints Type'
)
```

```python
top10 = df['Complaint Type'].value_counts().head(10)
others = pd.Series({'Other': df['Complaint Type'].value_counts().iloc[10:].sum()})
top10_with_others = pd.concat([top10, others])

top10_with_others.plot(kind='bar', figsize=(10,5), title='Top 10 Complaint Types + Other')
```
